Ok let's import packages and set up settings

```python
import numpy as np
import matplotlib.pyplot as plt

N = 200
xMax = 10
tau1 = 1
tau2 = 3
pTau1 = 0.6
irf_width = 0.2
noise_mag = 0.002
```

Now set up axes and base functions

```python
x = np.linspace(0,xMax,N)
data = pTau1*np.exp(-x/tau1) + (1-pTau1)*np.exp(-x/tau2)+noise_mag*np.random.randn(N)
data /= np.sum(data)
irf = np.zeros_like(x)
irf[np.where(x<irf_width)] = 1
irf /= np.sum(irf)
irf_noise = irf+noise_mag*np.random.randn(N)
conv = np.real_if_close(np.fft.ifft(np.fft.fft(irf_noise)*np.fft.fft(data)))
conv /= np.sum(conv)
print(np.shape(conv))
plt.plot(x,conv)
plt.plot(x,data)
plt.show()
```

Plot the data and irf separately

```python
plt.plot(x,data)
plt.semilogy()
plt.title("Double-exponential decay")
plt.show()
plt.figure()
plt.plot(x,irf_noise)
plt.title("Instrument response function")
plt.show()
plt.figure()
plt.plot(x,data, label = "original")
plt.plot(x,conv, label = "convolution")
plt.semilogy()
plt.legend()
plt.title("Data after convolution")
plt.show()
```

Now let's construct a noisy irf and background to use with our noisy data:

```python
irf_meas = irf + noise_mag*np.random.randn(N)
noise = noise_mag * np.random.randn(N)
backg = np.real_if_close(np.fft.ifft(np.fft.fft(irf_noise)*np.fft.fft(noise)))
plt.plot(x,irf_meas)
plt.plot(x,backg)
plt.show()
```

and build a covariance matrix:
$$\sigma_{ij} = \text{diag}\{\text{background}\} + \text{IRF}_{i-j}$$

```python
irfmat = np.tile(irf_meas,(N,1))
for i in range(N):
    irfmat[i,:] = np.roll(irfmat[i,:],i)
covar = np.diag(backg) + np.triu(irfmat)
plt.imshow(covar)
plt.show()
print("normalized log det: {}".format(np.round(np.log10(np.linalg.det(covar))/N,2)))
inv_cov = np.linalg.inv(covar)
plt.imshow(np.log10(np.abs(inv_cov)))
plt.colorbar()
plt.show()
```

Oops! Ok that doesn't look great so while I figure that out, let's set up a metropolis-hastings algorithm to maximize the likelihood

```python
def log_likelihood(params,disp = False):
    fit_tau1 = params[0]
    fit_tau2 = params[1]
    fit_ptau1 = params[2]
    expected = fit_ptau1*np.exp(-x/fit_tau1) + (1 - fit_ptau1)*np.exp(-x/fit_tau2)
    expected /= np.sum(expected)
    err = data - expected
    backg_smoothed = backg
    backg_smoothed[np.where(backg<0.01)]=0.01
    chisq = -(err/backg_smoothed)**2
    if disp:
        plt.plot(x,data,label = "observed")
        plt.plot(x,expected,label = "expected")
        plt.legend()
        plt.show()
        plt.plot(chisq)
        plt.show()
    if (fit_tau1<0) | (fit_tau2<0) | (fit_ptau1<0) | (fit_tau1>fit_tau2) | (fit_ptau1>1):
        return -np.inf
    return np.sum(chisq)
```

```python
init_param_guess = [0.1,2,0.5]
num_reps = 100000
jump_mag = [0.05,0.5,0.1]
param_its = np.zeros((3,num_reps))
param_its[:,0] = init_param_guess
rands = np.random.uniform(0,1,(num_reps))
for i in range(num_reps-1):
    my_likelihood = log_likelihood(param_its[:,i])
    jump = jump_mag * np.random.randn(np.size(init_param_guess))
    new_params = param_its[:,i] + jump
    new_likelihood = log_likelihood(new_params)
    if new_likelihood > my_likelihood:
        param_its[:,i+1] = new_params
    elif rands[i] <np.exp(new_likelihood - my_likelihood):
        param_its[:,i+1] = new_params
    else:
        param_its[:,i+1] = param_its[:,i]
plt.plot(np.arange(num_reps),param_its.T,".")
plt.show()
plt.hist(param_its[1,100:])
plt.show()
```
