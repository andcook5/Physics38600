Ok let's import packages and set up settings

```python
import numpy as np
import matplotlib.pyplot as plt

N = 200
xMax = 10
tau1 = 1
tau2 = 0.2
pTau1 = 0.6
irf_width = 2
```

Now set up axes and base functions

```python
x = np.linspace(0,xMax,N)
data = pTau1*np.exp(-x/tau1) + (1-pTau1)*np.exp(-x/tau2)
irf = np.zeros_like(x)
irf[np.where(np.abs(x-irf_width)<=irf_width/2)] = 1
irf /= np.sum(irf)
conv = np.convolve(data,irf,"same")
```

Plot the data and irf separately

```python
plt.plot(x,data)
plt.semilogy()
plt.title("Double-exponential decay")
plt.show()
plt.figure()
plt.plot(x,irf)
plt.title("Instrument response function")
plt.show()
plt.figure()
plt.plot(x,data, label = "original")
plt.plot(x,conv, label = "convolution")
plt.semilogy()
plt.legend()
plt.title("Data after convolution")
plt.show()
```
